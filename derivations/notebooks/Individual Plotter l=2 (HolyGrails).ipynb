```python
# Simple 2D plotter.
def plot_2d(n, r_ex, z_ex, u, uname):
    if n == 1:
        fig, ax = plt.subplots(figsize = (8, 8))
        im = ax.imshow(u.T, origin = 'lower', cmap = cm.viridis, extent = [r_ex[0], r_ex[1], z_ex[0], z_ex[1]])
        ax.set_xlabel(r"$\rho$")
        ax.set_ylabel(r"$z$")
        ax.set_title(uname)
        fig.colorbar(im, fraction=0.046, pad=0.04, ax = ax)
        
    if n > 1:
        fig = plt.figure(figsize = (16, ((n + 1) // 2) * 8))
        
        ax = []
        
        for i in range(n):
            ax.append( fig.add_subplot((n + 1) // 2, 2, i + 1))
            im = ax[-1].imshow(u[i].T, origin = 'lower', cmap = cm.viridis, extent = [r_ex[0], r_ex[1], z_ex[0], z_ex[1]])
            ax[-1].set_xlabel(r"$\rho$")
            ax[-1].set_ylabel(r"$z$")
            ax[-1].set_title(uname[i])
            plt.colorbar(im, fraction=0.046, pad=0.04, ax = ax[-1])
    
    plt.show()   
```

# Mass Calculation.

Some relevant variables.

```python
r_i = r[1:-1,1:-1]
z_i = z[1:-1,1:-1]
rr_i = np.sqrt(r_i**2 + z_i**2)

log_alpha = log_alpha_f[1:-1,1:-1]

log_a = log_a_f[1:-1,1:-1]
log_h = log_h_f[1:-1,1:-1]

Dr_log_a = 0.5 * (log_a_f[2:,1:-1] - log_a_f[0:-2,1:-1]) / dr
Dz_log_a = 0.5 * (log_a_f[1:-1,2:] - log_a_f[1:-1,0:-2]) / dz
Dr_log_h = 0.5 * (log_h_f[2:,1:-1] - log_h_f[0:-2,1:-1]) / dr
Dz_log_h = 0.5 * (log_h_f[1:-1,2:] - log_h_f[1:-1,0:-2]) / dz

# Radial derivatives.
D_rr_log_a = (r_i / rr_i) * Dr_log_a + (z_i / rr_i) * Dz_log_a
D_rr_log_h = (r_i / rr_i) * Dr_log_h + (z_i / rr_i) * Dz_log_h
```

## The Schwarzschild Pseudomass.

We can calculate the Schwarzschild pseudomass by calculating the area of 2-spheres.

$$M = \left(\frac{\tilde{A}}{16\pi}\right)^{1/2}\,\left(1 - \frac{(d\tilde{A}/dr)^2}{16\pi g_{rr} \tilde{A}}\right)\,.$$

Recall that our metric has the following form in spherical coordinates:

$$dl^2 = A^2\,dr^2 + r^2\,\left(A^2\,d\theta^2 + H^2\,\sin^2\theta\,d\varphi^2\right)\,.$$

Metric on 2-spheres is 

$$\Omega_{ab} = r^2\,\left(A^2\,d\theta^2 + H^2\,\sin^2\theta^2\,d\varphi^2\right)\,,$$

Thus,

$$\tilde{A} = \int_S\,\sqrt{\Omega}\,d\theta\,d\varphi = 2\pi\,r^2\,\int\limits_0^\pi\,AH\,\sin\theta\,d\theta\,.$$

By approximating that $A$ and $H$ do not have angular dependence far away, we can say that 

$$\tilde{A} = 4\pi\,r^2\,AH\,.$$

Thus, 

$$\frac{d \tilde{A}}{dr} = 8\pi\,AH\,r\,\left(1 + \frac{r}{2}\,\left(\partial_r \log A + \partial_r \log H\right)\right)\,.$$

$$M = \frac{r\,\sqrt{AH}}{2}\,\left(1 - \frac{H}{A}\,\left(1 + \frac{r}{2}\,\left(\partial_r \log A + \partial_r \log H \right)^2\right)\right)$$

```python
M_S = 0.5 * rr_i * np.exp(0.5 * (log_a + log_h)) * (1.0 - np.exp(log_h - log_a) * (1.0 + 0.5 * rr_i * (D_rr_log_a + D_rr_log_h))**2)
```

```python
fig, ax = plt.subplots(figsize = (8, 6))

ax.plot((rr_i[:,1])[rr_i[:,1] > 0.0], (M_S[:,1])[rr_i[:,1] > 0.0], label = r"$M_{S}(\rho)$")
ax.plot((rr_i[1,:])[rr_i[1,:] > 0.0], (M_S[1,:])[rr_i[1,:] > 0.0], label = r"$M_{S}(z)$")

ax.plot((np.diagonal(rr_i))[np.diagonal(rr_i) > 0.0], (np.diagonal(M_S))[np.diagonal(rr_i) > 0.0], label = r"$M_{S}(d)$")

ax.set_xlabel(r"$r$")
ax.set_ylabel(r"$M_S$")
ax.legend()
ax.set_title(r"Pseudo-Schwarzschild Mass $M_S$")
plt.show()
```

## ADM Mass.

We can use Gourgoulhon's expression, see eq. (4.20) in Gourgoulhon (2011).

$$M_{\text{ADM}} = -\frac{1}{16\pi}\,\lim_{\mathscr{S}\to\infty}\,\oint_{\mathscr{S}}\,\left[ \frac{\partial}{\partial r}\,(A^2 + H^2) + \frac{H^2 - A^2}{r}\right]\,r^2 \sin\theta\,d\theta\,d\varphi$$

By once again approximating that there is no angular dependence, we reach the expression

$$M_{\text{ADM}} = -\frac{1}{4}\,\left(2r^2\left(A^2\,\partial_r \log A + H^2\,\partial_r \log H\right) + r\,(H^2 - A^2)\right)$$

```python
M_ADM = -0.25 * (2.0 * rr_i**2 * (np.exp(2.0 * log_a) * D_rr_log_a + np.exp(2.0 * log_h) * D_rr_log_h) + rr_i * (np.exp(2.0 * log_h) - np.exp(2.0 * log_a)))
```

```python
fig, ax = plt.subplots(figsize = (8, 6))

ax.plot((rr_i[:,1])[rr_i[:,1] > 0.0], (M_ADM[:,1])[rr_i[:,1] > 0.0], label = r"$M_{ADM}(\rho)$")
ax.plot((rr_i[1,:])[rr_i[1,:] > 0.0], (M_ADM[1,:])[rr_i[1,:] > 0.0], label = r"$M_{ADM}(z)$")

ax.plot((np.diagonal(rr_i))[np.diagonal(rr_i) > 0.0], (np.diagonal(M_ADM))[np.diagonal(rr_i) > 0.0], label = r"$M_{ADM}(d)$")

ax.set_xlabel(r"$r$")
ax.set_ylabel(r"$M_{ADM}$")
ax.legend()
ax.set_title(r"ADM Mass $M_{ADM}$")
plt.show()
```

# Angular Momentum.

We can use Gourgoulhon's expression, see eq. (4.37) in Gourgoulhon (2011).

$$J = \frac{1}{16\pi}\,\lim_{\mathscr{S}\to\infty}\,\oint_{\mathscr{S}}\,\frac{\partial \beta^\varphi}{\partial r}\,r^4 \sin^3\theta\,d\theta\,d\varphi$$

Assuming as before that there is no angular dependence, we can get 

$$J = \frac{r^4}{6}\,\frac{\partial \beta^\varphi}{\partial r}\,.$$

```python
beta = beta_f[1:-1,1:-1]

Dr_beta = 0.5 * (beta_f[2:,1:-1] - beta_f[0:-2,1:-1]) / dr
Dz_beta = 0.5 * (beta_f[1:-1,2:] - beta_f[1:-1,0:-2]) / dz

D_rr_beta = (r_i / rr_i) * Dr_beta + (z_i / rr_i) * Dz_beta

J = D_rr_beta * rr_i**4 / 6.0
```

```python
fig, ax = plt.subplots(figsize = (8, 6))

ax.plot((rr_i[:,1])[rr_i[:,1] > 0.0], (J[:,1])[rr_i[:,1] > 0.0], label = r"$J(\rho)$")
ax.plot((rr_i[1,:])[rr_i[1,:] > 0.0], (J[1,:])[rr_i[1,:] > 0.0], label = r"$J(z)$")

ax.plot((np.diagonal(rr_i))[np.diagonal(rr_i) > 0.0], (np.diagonal(J))[np.diagonal(rr_i) > 0.0], label = r"$J(d)$")

ax.set_xlabel(r"$r$")
ax.set_ylabel(r"$J$")
ax.legend()
ax.set_title(r"Angular Momentum $J$")
plt.show()
```
