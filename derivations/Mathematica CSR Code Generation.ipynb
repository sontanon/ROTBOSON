````python
cc_code = """
void jacobian_4th_order_variable_omega_cc
(
	double *aa,		// CSR array for values.
	MKL_INT *ia, 		// CSR array for row beginnings. 
	MKL_INT *ja,		// CSR array for columns.
	const MKL_INT NrTotal, 	// Grid total dimension in r.
	const MKL_INT NzTotal, 	// Grid total dimension in z.
	const MKL_INT dim,	// Grid total 2D dimension: dim = NrTotal * NzTotal.
	const MKL_INT ghost,	// Number of ghost zones.
	const MKL_INT i, 	// Integer coordinate for r: 0 <= i < NrTotal.
	const MKL_INT j, 	// Integer coordinate for z: 0 <= j < NzTotal.
	const double dr, 	// Spatial step in r.
	const double dz,	// Spatial step in z.
	const MKL_INT l, 	// Scalar field rotation number.
	const double m, 	// Scalar field mass.
	const double xi,	// Scalar field frequency variable.
	// Now come the grid variables. For cc stencil, each grid function has 9 variables.
	const double u102, const double u112, const double u120, const double u121, const double u122, const double u123, const double u124, const double u132, const double u142,
	const double u202, const double u212, const double u220, const double u221, const double u222, const double u223, const double u224, const double u232, const double u242,
	const double u302, const double u312, const double u320, const double u321, const double u322, const double u323, const double u324, const double u332, const double u342,
	const double u402, const double u412, const double u420, const double u421, const double u422, const double u423, const double u424, const double u432, const double u442,
	const double u502, const double u512, const double u520, const double u521, const double u522, const double u523, const double u524, const double u532, const double u542,
	const double u602, const double u612, const double u620, const double u621, const double u622, const double u623, const double u624, const double u632, const double u642,
	const MKL_INT offset1,	// Number of elements filled before filling function 1.
	const MKL_INT offset2, 	// Number of elements filled before filling function 2.
	const MKL_INT offset3, 	// Number of elements filled before filling function 3.
	const MKL_INT offset4, 	// Number of elements filled before filling function 4.
	const MKL_INT offset5,	// Number of elements filled before filling function 5.
	const MKL_INT offset6	// Number of elements filled before filling function 5.
)
{
\t// Grid variables.
\tdouble u1 = u122;
\tdouble u2 = u222;
\tdouble u3 = u322;
\tdouble u4 = u422;
\tdouble u5 = u522;
\tdouble u6 = u622;
    
\t// Physical names for readability.
\tdouble alpha = exp(u1);
\tdouble Omega = u2;
\tdouble h = exp(u3);
\tdouble a = exp(u4);
\tdouble psi = u5;
\tdouble lambda = u6;
    
\t// Coordinates.
\tdouble ri = (double)i + 0.5 - ghost;
\tdouble r = ri * dr;
\tdouble r2 = r * r;
    
\t// Step ratios.
\tdouble dzodr = dz / dr;
\tdouble drodz = dr / dz;
\tdouble dr2 = dr * dr;
    
\t// Scalar field mass and frequency.
\tdouble w = omega_calc(xi, m);
\tdouble m2 = m * m;
\t// Omega variable index position.
\tMKL_INT w_idx = GNUM * dim;
    
\t// Short hands.
\t// Scalar field.
\tdouble rlm1 = (l == 1) ? 1.0 : pow(r, l - 1);
\tdouble rl = rlm1 * r;
\tdouble phior = rlm1 * psi;
\tdouble phi = r * phior;
\tdouble phi2or2 = phior * phior;
\tdouble phi2 = phi * phi;
\t// Shift combined with scalar field rotation and frequency.
\tdouble wplOmega = w + l * Omega;
\tdouble wplOmega2 = wplOmega * wplOmega;
\t// Squared variables.
\tdouble alpha2 = alpha * alpha;
\tdouble h2 = h * h;
\tdouble a2 = a * a;
\t// Regularization a2.
\tdouble a2_r = h2 + r2 * lambda;
    
\t// Finite differences.
\t// Axial derivatives.
\tdouble dRu1 = D10 * u102 + D11 * u112 + D13 * u132 + D14 * u142;
\tdouble dRu2 = D10 * u202 + D11 * u212 + D13 * u232 + D14 * u242;
\tdouble dRu3 = D10 * u302 + D11 * u312 + D13 * u332 + D14 * u342;
\t//double dRu4 = D10 * u402 + D11 * u412 + D13 * u432 + D14 * u442;
\tdouble dRu5 = D10 * u502 + D11 * u512 + D13 * u532 + D14 * u542;
\tdouble dRu6 = D10 * u602 + D11 * u612 + D13 * u632 + D14 * u642;
\t// Z derivatives.
\tdouble dZu1 = D10 * u120 + D11 * u121 + D13 * u123 + D14 * u124;
\tdouble dZu2 = D10 * u220 + D11 * u221 + D13 * u223 + D14 * u224;
\tdouble dZu3 = D10 * u320 + D11 * u321 + D13 * u323 + D14 * u324;
\t//double dZu4 = D10 * u420 + D11 * u421 + D13 * u423 + D14 * u424;
\tdouble dZu5 = D10 * u520 + D11 * u521 + D13 * u523 + D14 * u524;
\tdouble dZu6 = D10 * u620 + D11 * u621 + D13 * u623 + D14 * u624;
\t// Second derivatives.
\tdouble dRRu1 = D20 * u102 + D21 * u112 + D22 * u122 + D23 * u132 + D24 * u142;
\tdouble dRRu3 = D20 * u302 + D21 * u312 + D22 * u322 + D23 * u332 + D24 * u342;
    
\t// Declare Jacobian submatrices.
\tdouble jacobian_submatrix_1[5] = { 0.0 };
\tdouble jacobian_submatrix_2[5] = { 0.0 };
\tdouble jacobian_submatrix_3[5] = { 0.0 };
\tdouble jacobian_submatrix_4[5] = { 0.0 };
\tdouble jacobian_submatrix_5[5] = { 0.0 };
\tdouble jacobian_submatrix_6[5] = { 0.0 };
\tdouble jacobian_submatrix_w = 0.0;

"""
````

Now we have to generate CSR code for all six variables. The pseudocode goes something like this:

```C
// Write Jacobian matrix in question.
// First for log(alpha)
jacobian_submatrix_1[5] = { ... }; // Extract entries from Jacobian strings.
// Second for beta.
jacobian_submatrix_2[5] = { ... }; // Extract as well.
// Do for other variables.
// ...

// Write CSR arrays for grid number k.
// Index start array ia.
ia[(k - 1) * dim + IDX(i, j)] = BASE + offsetk;

// Now write values.
aa[offsetk +  0] = ... ; // Extract values.
// ... etc.

// Now columns.
ja[offsetk + 0] = ... ; // Calculate indices.
// ... etc.
```

````python
# Start main code.
main_code = """"""

# Loop over six grid functions.
for grid_num_1 in range(1,7):
    # For each grid number, start by writing which grid function we are dealing with.
    main_code += """\t// scR CODE FOR GRID NUMBER %d.\n\n""" % (grid_num_1)
    main_code += """\t// First write down Jacobian submatrices.\n"""
    
    # Write down submatrices. We have 6 submatrices for each grid function.
    for grid_num_2 in range(1, 7):
        # Write down which submatrix we are wrtiting.
        main_code += """\t// Submatrix %d.\n""" % (grid_num_2)
        
        # Loop over 5 variables: u, dRu, dZu, dRRu, dZZu.
        for i in range(5):
            main_code += """\tjacobian_submatrix_%d[%d] = %s;\n""" % (grid_num_2, i, jacobian[grid_num_1-1][5*(grid_num_2-1):5*grid_num_2][i])
        
        # Add newline for readability.
        main_code += """\n"""
    
    # Now that submatrices for grid functions have been written, add omega term.
    main_code += """\t// Omega term.\n"""
    if (jacobian[grid_num_1-1][-1] == "0" or jacobian[grid_num-1][-1] == "0.0"):
        main_code += """\tjacobian_submatrix_w = %s;\n""" % (jacobian[grid_num_1-1][-1])
    else:
        # Omega must be multiplied by chain rule du/d(xi) since xi is the real variable.
        main_code += """\tjacobian_submatrix_w = dw_du(xi, m) * (%s);\n""" % (jacobian[grid_num_1-1][-1]) 
    
    # Add newline for readability.
    main_code += """\n"""
    
    # Write ia term.
    main_code += """\t// This row %d * dim + IDX(i, j) starts at offset%d\n""" % (grid_num_1-1, grid_num_1)
    main_code += """\tia[%d * dim + IDX(i, j)] = BASE + offset%d;\n\n""" % (grid_num_1-1, grid_num_1)
    
    # Write values in aa and columns in ja.
    aa_code = """\t// Values.\n"""

    ja_code = """\t// Columns.\n"""

    # Nonzeros or offset counter.
    offset = 0

    # Loop over 6 grid variables.
    for i in range(6):

        # Loop over stencil points.
        for j in range(len(stencil_sc)):
            # Code string.
            aa_code_string = """"""
            ja_code_string = """"""

            # Loop over 5 stencil points.
            for k in range(5):
                # Stencil element.
                stencil_element = stencil_sc[j][k]
                # Jacobian element
                jacobian_element = jacobian[grid_num_1-1][5*i:5*(i+1)][k]

                # Check if stencil element is trivial.
                if (stencil_element != """0.0""") and (jacobian_element != """0"""):
                    # Write to list.
                    aa_code_string += """+""" + stencil_element + """*""" + """jacobian_submatrix_%d[%d]""" % (i + 1, k)

            if (aa_code_string != ""):
                # Add semicolon for line termination.
                aa_code_string += """;\n"""
                # Add scR matrix asignment.
                aa_code_string = """\taa[offset%d + %2d] = """ % (grid_num_1, offset) + aa_code_string

                # Now write ja string.
                ja_code_string = """\tja[offset%d + %2d] = BASE + %d * dim + IDX""" % (grid_num_1, offset, i) + indices_sc[j] + """;\n"""

                # Add to overall code.
                aa_code += aa_code_string
                ja_code += ja_code_string

                # Increase offset
                offset += 1
                
    # Add omega term if necessary.
    jacobian_element = jacobian[grid_num_1-1][-1]
    
    # Check if element is trivial.
    if (jacobian_element != """0"""):
        # Write to list.
        aa_code += """\taa[offset%d + %2d] = jacobian_submatrix_w;\n""" % (grid_num_1, offset)
        ja_code += """\tja[offset%d + %2d] = BASE + w_idx;\n""" % (grid_num_1, offset)
        
    # Add newlines for readability.
    aa_code += """\n"""
    ja_code += """\n\n"""

    # Add to main code.
    main_code += aa_code + ja_code
````

````python
# Add to cc_code and add also return statement.
sc_code += main_code + """\t// All done.\n\treturn;\n}"""
````
