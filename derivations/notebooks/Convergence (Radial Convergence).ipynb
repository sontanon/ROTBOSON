```python
from pylab import *
```

```python
Ns = np.linspace(125, 1000, 8)astype
```

```python
h = 125.0 / Ns
```

```python
print(Ns)
print(h)
```

```python
diff_M = np.array([0.013973005692740958, 
9.741797676676711e-05,
1.9538043805500394e-05, 
6.133736739855067e-06,
2.4713639188433945e-06])
```

```python
diff_J = np.array([0.05014167243366874,
0.002814818288441323,
0.000565897538349347,
0.00018069008330279246, 
7.431755116549255e-05])
```

```python
fig, ax = plt.subplots(figsize = (8, 6))
ax.plot(-np.log10(h[:5]), np.log10(diff_M[:5]), "-*", label=r"$|M_{Komar1} - M_{Komar2}|$")
ax.plot(-np.log10(h[:5]), (4.0 * (np.log10(h) - np.log10(h[1])) + np.log10(diff_M[1]))[:5] , label=r"$\mathcal{O}(h^4)$")
ax.set_xlabel(r"$\log_{10} (1/h)$")
ax.set_ylabel(r"$\log_{10} \Delta M$")
ax.set_title(r"Convergence of Komar Masses Difference")
ax.legend()
plt.show()
```

```python
fig, ax = plt.subplots(figsize = (8, 6))
ax.plot(-np.log10(h[:5]), np.log10(diff_J[:5]), "-*", label=r"$|J_{Komar1} - J_{Komar2}|$")
ax.plot(-np.log10(h[:5]), (4.0 * (np.log10(h) - np.log10(h[1])) + np.log10(diff_J[1]))[:5] , label=r"$\mathcal{O}(h^4)$")
ax.set_xlabel(r"$\log_{10} (1/h)$")
ax.set_ylabel(r"$\log_{10} \Delta J$")
ax.set_title(r"Convergence of Komar Angular Momenta Difference")
ax.legend()
plt.show()
```
