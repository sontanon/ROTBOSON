```python
base = "../Fixed Omega Comparison/"
```

```python
dirnames = os.listdir(base)[:]
dirnames
```

```python
r = np.genfromtxt(base + dirnames[0] + "/r.asc")
psi_0 = np.genfromtxt(base + dirnames[0] + "/phi_f.asc")

rr_l  = np.genfromtxt(base + dirnames[1] + "/sph_rr.asc", usecols=0)
psi_1 = np.genfromtxt(base + dirnames[1] + "/sph_psi_f.asc", usecols = 398)
psi_2 = np.genfromtxt(base + dirnames[2] + "/sph_psi_f.asc", usecols = 398)
psi_3 = np.genfromtxt(base + dirnames[3] + "/sph_psi_f.asc", usecols = 398)
psi_4 = np.genfromtxt(base + dirnames[4] + "/sph_psi_f.asc", usecols = 398)
psi_5 = np.genfromtxt(base + dirnames[5] + "/sph_psi_f.asc", usecols = 398)
psi_6 = np.genfromtxt(base + dirnames[6] + "/sph_psi_f.asc", usecols = 398)
```

```python
plt.rcParams.update({'font.size': 12})
```

```python
fig, ax = plt.subplots(figsize = (8, 6))
ax.plot(r, psi_0, label = r"$l = 0$")
ax.plot(rr_l, psi_1 * rr_l**1, label = r"$l = 1$")
ax.plot(rr_l, psi_2 * rr_l**2, label = r"$l = 2$")
ax.plot(rr_l, psi_3 * rr_l**3, label = r"$l = 3$")
ax.plot(rr_l, psi_4 * rr_l**4, label = r"$l = 4$")
ax.plot(rr_l, psi_5 * rr_l**5, label = r"$l = 5$")
ax.plot(rr_l, psi_6 * rr_l**6, label = r"$l = 6$")
ax.set_xlim(0,32)
ax.set_title(r"Scalar Field Profile At Equator $\phi(r,\,\theta = \pi/2)$ and $\omega = 0.8\,(m/\hbar)$")
ax.set_ylabel(r"$\phi$")
ax.set_xlabel(r"$r\,[\hbar/m]$")
ax.legend()
plt.show()
```
